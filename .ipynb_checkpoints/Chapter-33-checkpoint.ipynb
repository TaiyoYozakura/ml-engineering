### -> Loading Dataset

```python
import pandas as pd
data=pd.read_csv("student_scores.csv")
# data['timestamps'] = pd.to_datetime(data['timestamps'])
print(data.head(5))
```

## -> Data Preprocessing 

### ---> Inspecting the DataSet

```python
#Inspect columns and basic info

print('Columns', data.columns.tolist())

#1.Dataset Basic Info
print('\nDataset Info:')
print(data.info())

#2.Data Completeness
print('\nDataset Completeness:')
print(data.isnull().sum())
      
#3.Dataset Consistency
print('\nDataset Consistency:')
print(data.dtypes)
```

```python
#4. Data Accuracy
print("\nDataset Describe:")
data.describe()
      
#6. BIAS Check
print('\nDataset Bias')
print(data["gender"].value_counts (normalize=True))
print(data['grades'].value_counts(normalize=True))
data['class'].value_counts(normalize=True)
```

### -> Separating Features and Labels

```python
cat=data.select_dtypes(include='object').columns
numer=data.select_dtypes(exclude='object').columns
print("Categorical Features :",list(cat))
print("Numerical Features :",list(numer))

# Encoding Data :::: NOTE : Already did
# le=LabelEncoder()
# data["gender"]=le.fit_transform(data["gender"])
# print(data.head(5))
# data["class_en"]=le.fit_transform(data["grades"])

x=data[["hours"]]
y=data['score']
```

### -> Data Distribution

```python
import matplotlib.pyplot as mpl

mpl.figure(figsize=(6,4))
mpl.hist(x,bins=20,color='skyblue',edgecolor='black')
mpl.xlabel('Hours')
mpl.ylabel('Frequency')
mpl.title('Distribution of Student Hours.')
mpl.show()
    
mpl.figure(figsize=(6,4))
mpl.hist(y,bins=20,color='skyblue',edgecolor='black')
mpl.xlabel('Scores')
mpl.ylabel('Frequency')
mpl.title('Distribution of Student Scores.')
mpl.show()
```

### -> Spliting the Data

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=84)
# additional parameter stratify=y
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
print("Model Coefficient (Slope) :",model.coef_[0])
print("Model Intercept :",model.intercept_)
```

### -> Make Predictions For Test Data

```python
y_pred=model.predict(x_test)
user_input=float(input("Enter Number of Hours Studied : "))
print("You can Score",model.predict([[user_input]]),"after studying for",user_input,"hrs.")
```
