### Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
X=[[1], [2], [3], [4]]
y = [2, 4, 6, 8]
model.fit(X, y)
prediction=model.predict([[10]])
print("Prediction for input 5 is:", prediction)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
x=[[1],[2],[3],[4]]
y=[2,4,6,8]
model.fit(x,y)
user_val=float(input("Enter a number to predict : "))
pred=model.predict([[user_val]])
print("Prediction is :",pred)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y,color="blue",label="Data Points")
plt.plot(x,model.predict(x),color="orange",label='Regression Line')
plt.legend(loc='center left',bbox_to_anchor=(1,0.5))
plt.title("Linear Regression Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout()
plt.show()
```

### Activity 1 Banana Price-Day Prediction Model

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
d=[[1],[2],[3],[4],[5]]
p=[12,20,26,30,38]
model.fit(d,p)
user_val=float(input("Enter a Day to predict banana Price for : "))
pred=model.predict([[user_val]])
print("Predicted Price is :",pred)
```

```python
import matplotlib.pyplot as plt
plt.scatter(d,p,color="blue",label="Data Points")
plt.plot(d,model.predict(d),color="orange",label='Regression Line')
plt.legend(loc='center left',bbox_to_anchor=(1,0.5))
plt.title("Linear Regression Line")
plt.xlabel("Day")
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```

### Activity 2 Study Hours-Marks Scored Prediction Model

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
h=[[1],[2],[3],[4],[5],[6]]
m=[35,40,45,50,55,60]
model.fit(h,m)
user_val=float(input("Enter a number to predict : "))
pred=model.predict([[user_val]])
print("Prediction is :",pred)
```

```python
import matplotlib.pyplot as plt
plt.scatter(h,m,color="blue",label="Data Points")
plt.plot(h,model.predict(h),color="orange",label='Regression Line')
plt.legend(loc='center left',bbox_to_anchor=(1,0.5))
plt.title("Linear Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks can Score")
plt.tight_layout()
plt.show()
```
