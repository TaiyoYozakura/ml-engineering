```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve,auc,confusion_matrix,classification_report
import matplotlib.pyplot as pt
import numpy as np
```

```python
# Data separation
x,y=make_classification(n_samples=1000,n_classes=2,random_state=8)
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=84)
# model training
model=LogisticRegression()
model.fit(xtrain,ytrain)
# probabilities
yprobs=model.predict_proba(xtest)[:,1]
print(yprobs)
```

```python
# ROC+AUC
fpr,tpr,threshold=roc_curve(ytest,yprobs)
roc_auc=auc(fpr,tpr)
print("AUC score:",roc_auc)
# finding best threshold value
jscores=tpr-fpr
bestidx=np.argmax(jscores)
bestThres=threshold[bestidx]
print("Best Threshold:",bestThres)
# apply best threshod
ypredCustom=(yprobs>=bestThres).astype(int)
print("\nConfusion Matrix:\n",confusion_matrix(ytest,ypredCustom))
print("\nClassification Report:\n",classification_report(ytest,ypredCustom))
```

```python
pt.figure(figsize=(8,6))
pt.plot(fpr,tpr,label=f"ROC curve(AUC={roc_auc:.2f})")
pt.plot([0,1],[0,1],linestyle="--")
step=max(1,len(threshold)//10)
for i in range(0,len(threshold),step):
    pt.scatter(fpr[i],tpr[i],color="red")
    pt.text(fpr[i],tpr[i],f"{threshold[i]:.2f}",fontsize=8)
pt.scatter(fpr[bestidx],tpr[bestidx],color="blue",s=120,
            label=f"Best Threshold={bestThres:.2f}")
pt.xlabel("False Positive Rate")
pt.ylabel("True Positive Rate")
pt.title("ROC curve with Thresholds")
pt.legend(loc="lower right")
pt.grid()
pt.show()
```
