```python
import streamlit as st
from joblib import load
import pandas as pd

from streamlit_jupyter import StreamlitPatcher

# Patch streamlit to work in Jupyter
StreamlitPatcher().jupyter()
```

```python
model=load("pipeline.joblib")
```

```python
st.title("Student Pass Predicter")
hrs = st.number_input("Study Hours", min_value=0.0, step=0.5)
attendance = st.number_input("Attendance (%)", min_value=0.0, max_value=100.0, step=1.0)
```

```python
if st.button("Predict"):
    x = pd.DataFrame({
        "studyHours": [hrs],
        "attendance": [attendance]
    })
    prediction = model.predict(x)
    
    if prediction[0] == 1:
        st.success("Prediction : PASS")
    else:
        st.error("Prediction : FAIL")
```
