# -- Logistic Regression With Pipelines and JobLib --

#### Import Statements

```python
import pandas as pd
from joblib import dump,load
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
```

#### Loading Dataset and Separating Features and Labels 

```python
data=pd.DataFrame({
    "studyHours":[1,2,3,4,5,6,7,8],
    "attendance":[55,60,65,70,75,80,85,90],
    "pass":[0,0,0,0,1,1,1,1]
})

x=data[["studyHours","attendance"]]
y=data["pass"]
```

#### Data Split- Train and Test Data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=84)
```

#### Creating Pipeline with StandardScaler and Logistic Regression Model

```python
def nullHandler(data):
    print(data.isnull().sum())
```

```python
pipeline=Pipeline([
    # ("nullHandler",nullHandler(data)),
    ("scalar",StandardScaler()),
    ("model",LogisticRegression())
])
```

#### Training Data on Pipeline

```python
pipeline.fit(x_train,y_train)
yPred=pipeline.predict(x_test)
print(f"Accuracy : {accuracy_score(y_test,yPred)}")
print(f"Confusion Matrix : \n{confusion_matrix(y_test,yPred)}")
```

#### Creating Dumps of Pipeline, Model and Scaler

```python
dump(pipeline,"pipeline.joblib")
dump(pipeline.named_steps["scalar"],"scalar.joblib")
dump(pipeline.named_steps["model"],"model.joblib")
print()
```

#### Prediction Method 1 : Making Predictions Using Pipeline

```python
# studyHrs=int(input("Enter Study Hours : "))
# attendance=int(input("Enter Your Attendance : "))
studyHrs=6
attendance=67
newData=pd.DataFrame({
    "studyHours":[studyHrs],
    "attendance":[attendance]})
pipeline_loaded=load("pipeline.joblib")
pred=pipeline_loaded.predict(newData)
print("Pipeline Prediction :","Pass" if pred[0]== 1 else "Fail")
```

#### Option 2 :Making Predictions using Model Only

```python
scalerLoaded=load("scalar.joblib")
modelLoaded=load("model.joblib")
newDataScaled=scalerLoaded.transform(newData)
prediction=modelLoaded.predict(newDataScaled)
print("Model-Only Prediction :","Pass" if prediction[0]==1 else "Fail")
```
