```python
# PCA--> Principal Component Analysis
# its a column consisting of some parts of the each input feature column value for a specific row
# its primarily used to reduce the number of columns used to train the model to reduce the computation required in training of the model
# steps to calculate PCA
# 1.calculate mean of each column
# 2.create a new table(X_centered) with each column in original data set where values in each row as Xi-Xi_mean where Xi is the column (note roundoff
#     decimal values upto 3 places after decimal point
# 3.using covariance formula cov(x1,x2)=summation of { (x1*x2) } / (n-1) 
# 4.new table will be formed named Covariance table 
# 5.using eigen values and vectors on the X_centered Table or Data only if its square matrix
```

```python
import numpy as np
A = np.array([
   [3.5, 3.4, 6.2, 6.3], 
    [3.4, 7.9, 4.7, 10],
    [6.2, 4.7, 11.9, 10],
    [6.3, 10, 10, 14.7]
])

egval,egvec=np.linalg.eig(A)
print("Eigen Values:\n",egval)
print("Eigen Vectors:\n",egvec)
```

```python
# imports
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
#Loading Dataset
data=pd.read_csv("Advertising.csv")
```

```python
#preprocessing and data separation
x=data.iloc[:,:-1]
y=data.iloc[:,-1]
#data spliting
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=84)
sc=StandardScaler()
xtrain=sc.fit_transform(xtrain)
xtest=sc.transform(xtest)
```

```python
# Using PCA
pca=PCA(n_components=2)
xtrain=pca.fit_transform(xtrain)
xtest=pca.transform(xtest)
#convert pca results to dataframe
pca_train=pd.DataFrame(xtrain,columns=["PCA1","PCA2"])
pca_test=pd.DataFrame(xtest,columns=["PCA1","PCA2"])
print("PCA for Train:\n",pca_train)
print("PCA for Test:\n",pca_test)
```

```python
#Model Training
model=LinearRegression()
model.fit(pca_train,ytrain)
#Predicting for test data
ypred=model.predict(pca_test)
#Mse and r2 score calulation
mse=mean_squared_error(ytest,ypred)
r2=r2_score(ytest,ypred)
print("Mean Squared Error:\n",mse)
print("R2 score :\n",r2)
```
